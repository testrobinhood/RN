```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
url = "housing.csv"

df = pd.read_csv(url)

print(df.head())
```

```python
plt.plot(df["median_house_value"].head(20), marker="o")

plt.title("House Prices")
plt.xlabel("House Number")
plt.ylabel("Median House Value")
plt.show()
```

```python
avg_price = df.groupby("ocean_proximity")[
    "median_house_value"
].mean()

plt.bar(avg_price.index, avg_price.values)

plt.title("Average House Price by Location")
plt.xlabel("Ocean Proximity")
plt.ylabel("Average House Value")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.hist(df["median_house_value"], bins=20, edgecolor="black")

plt.title("Distribution of House Prices")
plt.xlabel("House Value")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.scatter(
    df["median_income"],
    df["median_house_value"],
    alpha=0.5
)

plt.title("Income vs House Value")
plt.xlabel("Median Income")
plt.ylabel("Median House Value")
plt.show()
```

```python
location_counts = df["ocean_proximity"].value_counts()

plt.pie(
    location_counts.values,
    labels=location_counts.index,
    autopct="%1.1f%%"
)

plt.title("Housing Records by Location Category")
plt.show()
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.hist(df["median_house_value"], bins=20)
plt.title("House Price Distribution")

plt.subplot(2, 2, 2)
plt.scatter(df["median_income"], df["median_house_value"])
plt.title("Income vs House Value")

plt.subplot(2, 2, 3)
plt.plot(df["median_house_value"].head(20))
plt.title("First 20 House Values")

plt.subplot(2, 2, 4)
avg_price.plot(kind="bar")
plt.title("Average Price by Location")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["median_house_value"], bins=20)

plt.title("House Price Distribution")
plt.xlabel("House Value")
plt.ylabel("Frequency")

plt.savefig("house_prices.png")
plt.show()
```
