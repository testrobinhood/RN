```python
# Document Retrieval Using Inverted Index
```

```python
documents = {
    1: "python is easy",
    2: "python is useful",
    3: "data science is useful"
}
```

```python
# Create inverted index
index = {}

for id, text in documents.items():
    for word in text.split():
        if word not in index:
            index[word] = []
        index[word].append(id)

print("Inverted Index:")
print(index)
```

```python
# Search a word
word = input("Enter word to search: ")

if word in index:
    print("Documents found:", index[word])
else:
    print("Word not found")
```
