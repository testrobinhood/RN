```python
# PageRank Algorithm

import networkx as nx
```

```python
# Step 1: Create a web page graph
G = nx.DiGraph()

G.add_edges_from([
    ('A', 'B'),
    ('A', 'C'),
    ('B', 'C'),
    ('C', 'A'),
    ('D', 'C')
])
```

```python
# Step 2: Calculate PageRank
rank = nx.pagerank(G)
```

```python
# Step 3: Display PageRank
print("PageRank of each page:")

for page, score in rank.items():
    print(page, ":", round(score, 3))
```
