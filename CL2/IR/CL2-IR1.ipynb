```python
import re
from nltk.stem import PorterStemmer
```

```python
text = "Information retrieval systems retrieve relevant documents from large collections. These systems are searching documents."
```

```python
words = re.findall(r'\b[a-zA-Z]+\b', text.lower())
print(words)
```

```python
stop_words = {"is", "are", "the", "from", "and", "a", "an", "in", "of"}

filtered = [w for w in words if w not in stop_words]
print(filtered)
```

```python
ps = PorterStemmer()

stemmed = [ps.stem(w) for w in filtered]
print(stemmed)
```

```python
print("Original Text:", text)
print("After Stopword Removal:", filtered)
print("After Stemming:", stemmed)
```
