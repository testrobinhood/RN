```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
G = nx.DiGraph()

G.add_edges_from([
    ("A", "B"),
    ("A", "C"),
    ("B", "C"),
    ("C", "A"),
    ("D", "C")
])
```

```python
pagerank = nx.pagerank(G, alpha=0.85)
```

```python
print("PageRank Scores:")

for page, score in sorted(pagerank.items()):
    print(f"Page {page}: {score:.4f}")
```

```python
plt.figure(figsize=(7, 5))

pos = nx.spring_layout(G, seed=7)

nx.draw_networkx(
    G,
    pos,
    with_labels=True,
    node_size=1800,
    font_size=12,
    arrows=True,
    arrowsize=20
)

plt.title("Web Page Link Network")
plt.axis("off")
plt.show()
```
