```python
import re
from collections import defaultdict
```

```python
documents = {
    1: "information retrieval is important",
    2: "retrieval systems search documents",
    3: "information systems store data",
    4: "search engines retrieve relevant information"
}
```

```python
inverted_index = defaultdict(list)

for doc_id, content in documents.items():
    words = set(re.findall(r"\b[a-zA-Z]+\b", content.lower()))

    for word in words:
        inverted_index[word].append(doc_id)
```

```python
for word in sorted(inverted_index):
    print(word, ":", inverted_index[word])
```

```python
def search(query):
    terms = re.findall(r"\b[a-zA-Z]+\b", query.lower())

    if not terms:
        return []

    result = set(inverted_index.get(terms[0], []))

    for term in terms[1:]:
        result = result.intersection(inverted_index.get(term, []))

    return sorted(result)
```

```python
query = "information search"

print("Query:", query)
print("Matching document IDs:", search(query))
```
