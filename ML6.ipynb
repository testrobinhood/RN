```python
import random
```

```python
#Setting up the environment
board = [" "] * 9
Q = {}
alpha = 0.5
gamma = 0.9
epsilon = 0.2
```

```python
# Defining the Tic-Tac-Toe game
def show_board(board):
    print()
    for i in range(0, 9, 3):
        print("|".join(board[i:i+3]))
        if i < 6:
            print("-" * 5)

def check_winner(b, player):
    wins = [
        (0,1,2), (3,4,5), (6,7,8),
        (0,3,6), (1,4,7), (2,5,8),
        (0,4,8), (2,4,6)
    ]
    return any(all(b[i] == player for i in line)
               for line in wins)

def get_state(b):
    return "".join(b)

def available_moves(b):
    return [i for i in range(9) if b[i] == " "]
```

```python
# Building the reinforcement learning model
def choose_move(state, moves):
    if random.random() < epsilon:
        return random.choice(moves)

    values = [Q.get((state, m), 0) for m in moves]
    best = max(values)
    choices = [m for m in moves if Q.get((state, m), 0) == best]
    return random.choice(choices)
```

```python
# Training the model
def train(episodes=30000):
    global epsilon

    for _ in range(episodes):
        b = [" "] * 9
        player = "X"
        history = []

        while True:
            state = get_state(b)
            moves = available_moves(b)

            if player == "X":
                move = choose_move(state, moves)
                history.append((state, move))
            else:
                move = random.choice(moves)

            b[move] = player

            if check_winner(b, player):
                reward = 1 if player == "X" else -1
                break

            if not available_moves(b):
                reward = 0
                break

            player = "O" if player == "X" else "X"

        # Update Q-values for X's moves
        for state, move in reversed(history):
            old = Q.get((state, move), 0)
            Q[(state, move)] = old + alpha * (reward - old)
            reward *= gamma

    epsilon = 0
```

```python
# Testing the model
def play_game():
    b = [" "] * 9
    print("Positions:")
    print("0|1|2\n3|4|5\n6|7|8")

    while True:
        show_board(b)

        move = int(input("Enter your move (0-8): "))
        if move not in available_moves(b):
            print("Invalid move! Try again.")
            continue

        b[move] = "O"

        if check_winner(b, "O"):
            show_board(b)
            print("You win!")
            return

        if not available_moves(b):
            show_board(b)
            print("Draw!")
            return

        state = get_state(b)
        move = choose_move(state, available_moves(b))
        b[move] = "X"
        print("AI chooses:", move)

        if check_winner(b, "X"):
            show_board(b)
            print("AI wins!")
            return

        if not available_moves(b):
            show_board(b)
            print("Draw!")
            return
```

```python
# Train and play
train()
print("Training completed!")
play_game()
```
